```python
import torch
import os
import pandas as pd

emb_dir = "../dimenet10conf_embeddings"
# emb_dir = "./" 

records = []

for fname in sorted(os.listdir(emb_dir)):
    if fname.endswith(".pt"):
        path = os.path.join(emb_dir, fname)
        try:
            x = torch.load(path, map_location="cpu")
            shape = tuple(x.shape)
            records.append({
                "filename": fname,
                "shape": str(shape),
                "num_samples": shape[0] if len(shape) > 0 else None,
                "embedding_dim": shape[1] if len(shape) > 1 else None,
                "filesize_MB": round(os.path.getsize(path) / 1024 / 1024, 2)
            })
        except Exception as e:
            records.append({
                "filename": fname,
                "shape": "Error",
                "num_samples": None,
                "embedding_dim": None,
                "filesize_MB": None,
                "error": str(e)
            })

df = pd.DataFrame(records)
display(df)
```
