```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Define colors and theme
colors = ["#004488", "#DDAA33", "#BB5566", "#000000"]
color = sns.color_palette(colors)
plt.rcParams.update({"figure.figsize": (10, 4)})
sns.set(rc={"figure.dpi": 200})
sns.set_theme(style="ticks", palette=color)
plt.tight_layout()

# Create dataframe for plotting
metrics_df = pd.DataFrame(
    {
        "Model": [
            "No Conformers",
            "No Conformers",
            "10-Fold",
            "10-Fold",
            "20-Fold",
            "20-Fold",
            "100-Fold",
            "100-Fold",
        ],
        "Metric": [
            "Literature R²",
            "Experimental R²",
            "Literature R²",
            "Experimental R²",
            "Literature R²",
            "Experimental R²",
            "Literature R²",
            "Experimental R²",
        ],
        "Value": [
            0.61,
            0.65,
            0.75,
            0.74,
            0.72,
            0.62,
            0.72,
            0.66,
        ],
    }
)

# Create the plot
ax = sns.barplot(data=metrics_df, x="Model", y="Value", hue="Metric")
ax.bar_label(ax.containers[0], fmt="%.2f", fontsize=10)
ax.bar_label(ax.containers[1], fmt="%.2f", fontsize=10)
plt.tight_layout()
plt.legend(title="", loc="upper left", fontsize=10)
plt.title("Model Performance", fontsize=12)
plt.ylim(0, 1.0)
plt.show()
```
