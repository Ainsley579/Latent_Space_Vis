```python
from markboard_v1 import MarkBoard
import numpy as np

x = np.random.randn(200)
y = np.random.randn(200)

mb = MarkBoard(x, y)
mb.show()
```

```python
import plotly.express as px
from markboard import MarkBoard  

df = px.data.iris()
fig = px.scatter(df, x="sepal_length", y="petal_width")

board = MarkBoard.attach(fig)   
board.show()                    
```

```python
fig = px.histogram(df, x="sepal_length")
board = MarkBoard.attach(fig)
board.show()
```

```python
import plotly.express as px
from markboard import MarkBoard

df = px.data.iris()
fig = px.scatter(df, x="sepal_length", y="petal_width", color="species")
fig.update_layout(title="Test MarkBoard on scatter")

board = MarkBoard.attach(fig)
board.show()
```
