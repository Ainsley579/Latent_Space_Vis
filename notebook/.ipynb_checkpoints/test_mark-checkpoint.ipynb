```python
import numpy as np
import plotly.graph_objs as go
from plotly.subplots import make_subplots
from ipywidgets import VBox, HBox, Button, ColorPicker, Text, Output
from IPython.display import display

# 1. 造一点假数据
np.random.seed(0)
N = 200
x = np.random.randn(N)
y = np.random.randn(N)

# 初始化颜色：全部灰色
colors = ["lightgray"] * N

# 2. 建一个 FigureWidget，这样 selection 事件可以在 Python 里接到
fig = go.FigureWidget(
    data=[
        go.Scatter(
            x=x,
            y=y,
            mode="markers",
            marker=dict(size=8, color=colors)
        )
    ],
    layout=go.Layout(
        title="Interactive Marking Demo",
        dragmode="select"  # 用 box select，也可以改成 'lasso'
    )
)

scatter = fig.data[0]

# 3. 存放 mark 信息的列表
marks = []
mark_counter = 0

# 4. 交互控件：颜色选择 + 文本标签 + 按钮
color_picker = ColorPicker(
    description='Color',
    value='#e41a1c'  # 默认红色
)

label_text = Text(
    description='Label',
    value='mark-1',
    placeholder='Optional mark name'
)

create_btn = Button(
    description='Create Mark from Selection',
    button_style='success'
)

info_out = Output()

last_selected_indices = []  # 最近一次 selection 里的点索引


# 5. 当用户在图上选点时的回调
def handle_selection(trace, points, selector):
    global last_selected_indices
    last_selected_indices = points.point_inds  # 这是一个索引列表
    with info_out:
        print(f"Selected {len(last_selected_indices)} points.")


scatter.on_selection(handle_selection)


# 6. 当用户点击“Create Mark”按钮时的回调
def handle_create_mark(btn):
    global mark_counter, colors
    
    if not last_selected_indices:
        with info_out:
            print("No points selected. Please select some points first.")
        return
    
    mark_counter += 1
    mark_id = mark_counter
    label = label_text.value if label_text.value.strip() else f"mark-{mark_id}"
    color = color_picker.value
    
    # 更新全局 colors 数组
    for idx in last_selected_indices:
        colors[idx] = color
    
    # 更新图上的颜色
    with fig.batch_update():
        scatter.marker.color = colors
    
    # 记录这个 mark
    marks.append({
        "id": mark_id,
        "label": label,
        "color": color,
        "indices": list(last_selected_indices)
    })
    
    # 输出当前 mark 状态
    with info_out:
        print(f"Created mark #{mark_id} ({label}) with color {color}, "
              f"{len(last_selected_indices)} points.")
        print("Current marks:", [m["label"] for m in marks])


create_btn.on_click(handle_create_mark)


# 7. 展示界面
ui = VBox([
    fig,
    HBox([color_picker, label_text, create_btn]),
    info_out
])

display(ui)
```
